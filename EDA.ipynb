#Missing Values

```python
df.isnull().sum()
```

```python
features_with_missing_values = []

for col in df.columns:
  if df[col].isnull().sum() > 0:
    features_with_missing_values.append(col)
```

```python
print(features_with_missing_values)
```

```python
features_with_missing_values_dic = {}

for col in features_with_missing_values:
  features_with_missing_values_dic[col] = df[col].isnull().sum()
```

```python
print(features_with_missing_values_dic)
```

#Visulaizing
